## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

## Data Wrangling

### Gathering Data

#### Load df ...

```python
import pandas as pd
import glob

# Ambil semua file csv di dalam folder PRSA
files = sorted(glob.glob('data/PRSA_Data_20130301-20170228/*csv'))
print(f"Jumlah File: {len(files)}")

#Gabungkan semua jadi satu dataframe
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)

#Cek hasilnya
print(df.shape)
df.head()
```

**Insight:** (Opsional)

Data berhasil digabungkan dari 12 file CSV menjadi satu dataframe dengan 420.768 baris dan 18 kolom. Data ini mencakup data kualitas udara dari 12 stasiun pemantau di Beijing selama periode Maret 2013 - Februari 2017, dengan variabel polutan (PM2.5, PM10, SO2, NO2, O3, CO) dan variabel cuaca (TEMP, PRES, DEWP, RAIN, wd, WSPM).


### Assessing Data

#### Identifying Missing Values and Duplicates

```python
# 1. Cek tipe data dan jumlah data
df.info

# 2. Cek data duplikat
print("\nMissing Value:")
print(df.isnull().sum())

# 3. Cek data duplikat
print(f"\nJumlah duplikat: {df.duplicated().sum()}")

# 4. Cek statistik deskriptif
df.describe()
```

**Steps to Take:**

- Menangani missing value pada kolom polutan (PM2.5, PM10, SO2, NO2, CO, O3) dan kolom cuaca (TEMP, PRES, DEWP, RAIN, wd, WSPM) dengan mengisi menggunakan nilai median/mode.
- Menangani outlier pada PM2.5 dan PM10 yang mencapai 999 karena nilainya jauh dari median (55 dan 82).
- Menggabungkan kolom year, month, day, hour menjadi satu kolom datetime untuk analisis tren waktu.
- Tidak perlu menghapus data duplikat karena jumlah duplikat = 0.

**Insight:** (Opsional)
- Terdapat missing value yang besar pada kolom polutan, terutama CO, O3, dan NO2, serta missing value kecil pada kolom cuaca.
- Nilai maksimum PM2.5 dan PM10 mencapai 999 yang jauh dari mediannya (55 dan 82), menunjukkan adanya outlier.
- Tidak ditemukan data duplikat, sehingga tahap cleaning berfokus pada penanganan missing value, outlier, dan pembuatan kolom datetime.

### Cleaning Data

#### Fixing Missing Values and Outliers

```python
# Copy dataframe biar data asli aman
df_clean = df.copy()

# 1. Isi missing value kolom numerik pakai median
for col in ['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# 2. Isi missing value kolom wd(teks) pakai modus / nilai yang sering muncul
df_clean['wd'] = df_clean['wd'].fillna(df_clean['wd'].mode()[0])

# 3. Gabungkan year, month, day, hour jadi 1 kolom datetime
df_clean['datetime'] =pd.to_datetime(df_clean[['year', 'month', 'day', 'hour']])

# 4. Cek hasil cleaning
print("sisa missing value setelah cleaning")
print(df_clean.isnull().sum())
print(f"\nJumlah baris: {df_clean.shape[0]}, Jumlah kolom: {df_clean.shape[1]}")

contoh= df_clean['datetime'].iloc[0]
print("kolom baru datetime berhasil dibuat, contoh",contoh)
```

**Insight:** (Opsional)
- Missing value pada kolom polutan dan cuaca berhasil diatasi dengan mengisi menggunakan nilai median untuk kolom numerik dan modus untuk kolom wd, sehingga tidak ada lagi data kosong.
- Outlier berupa nilai 999 pada PM2.5 dan PM10 yang merupakan error sensor berhasil diatasi dengan menggantinya menggunakan nilai median.
- Kolom datetime berhasil dibuat dari gabungan year, month, day, dan hour untuk memudahkan analisis tren waktu.
- Data kini bersih dan siap digunakan untuk tahap Exploratory Data Analysis (EDA) dengan 420.768 baris dan 19 kolom

## Exploratory Data Analysis (EDA)

### Explore Descriptive Statistics and Correlation

```python
# 1. Statistik deskriptif
print("Statistik Deskriptif:")
print(df_clean.describe().T)

# 2. Rata-rata polutan utama
print("\nRata-rata polutan:")
print(df_clean[['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3']].mean())

# 3. Korelasi antar variabel
print("\nKorelasi dengan PM2.5:")
korelasi = df_clean[['PM2.5', 'PM10', 'SO2', 'NO2', 'CO', 'O3', 'TEMP', 'PRES', 'DEWP', 'RAIN', 'WSPM']].corr()
print(korelasi['PM2.5'].sort_values(ascending=False))
```

**Insight:** (Opsional)
- Rata-rata PM2.5 sebesar 79.28 dan PM10 sebesar 104.26 menunjukkan kualitas udara di Beijing secara umum berada pada kategori tidak sehat selama periode 2013-2017.
- PM10 memiliki korelasi paling kuat dengan PM2.5 sebesar 0.88, artinya kedua partikel ini naik-turun bersamaan karena berasal dari sumber yang sama.
- CO (0.77) dan NO2 (0.66) juga berkorelasi kuat dengan PM2.5, menandakan emisi kendaraan dan industri adalah kontributor utama polusi partikel halus.
- WSPM (kecepatan angin) berkorelasi negatif sebesar -0.27, artinya semakin kencang angin, semakin rendah PM2.5 karena polutan tersebar.
- RAIN (-0.01) hampir tidak berpengaruh terhadap PM2.5, sedangkan O3 (-0.15) berkorelasi negatif karena memiliki pola pembentukan yang berbeda dari polutan partikel.

## Visualization & Explanatory Analysis

### Pertanyaan 1:
Bagaimana tren konsentrasi PM2.5 dan PM10 di Beijing dari tahun 2013 sampai 2017?

```python
plt.figure(figsize=(12, 6))
yearly = df.groupby('year')[['PM2.5', 'PM10']].mean()

plt.plot(yearly.index, yearly['PM2.5'], marker='o', label='PM2.5')
plt.plot(yearly.index, yearly['PM10'], marker='s', label='PM10')

plt.title('Tren Rata-rata PM2.5 dan PM10 di Beijing (2013-2017)')
plt.xlabel('Tahun')
plt.ylabel('Konsentrasi')
plt.legend()
plt.grid(True)
plt.show()
```

### Pertanyaan 2:
Faktor cuaca apa saja (suhu, kelembapan, kecepatan angin, tekanan udara) yang paling berpengaruh terhadap tingginya PM2.5?

```python
plt.figure(figsize=(9, 6))
cols = ['PM2.5', 'TEMP', 'DEWP', 'PRES', 'WSPM', 'RAIN']
sns.heatmap(df[cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Korelasi PM2.5 dengan Faktor Cuaca')
plt.show()
```

**Pertanyaan 3**
Stasiun pemantau mana yang memiliki kualitas udara paling buruk dan pada musim apa polusi paling tinggi?

```python
def get_season(month):
    if month in [12, 1, 2]:
        return 'Dingin'
    elif month in [3, 4, 5]:
        return 'Semi'
    elif month in [6, 7, 8]:
        return 'Panas'
    else:
        return 'Gugur'

df['season'] = df['month'].apply(get_season)

plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='station', y='PM2.5', hue='season', errorbar=None)
plt.title('Rata-rata PM2.5 per Stasiun dan Musim')
plt.xticks(rotation=30)
plt.show()
```

**Insight:** (Opsional)
- Musim Dingin memiliki rata-rata PM2.5 tertinggi di semua stasiun, sedangkan Musim Panas paling rendah.
- Stasiun dengan kualitas udara terburuk adalah Wanshouxigong, dengan PM2.5 musim dingin mencapai ∼108, tertinggi di antara semua stasiun.
- Stasiun dengan kualitas udara paling baik adalah Dingling, dengan PM2.5 paling rendah di semua musim.
- Jadi, polusi paling parah terjadi saat musim dingin dan paling terpusat di daerah Wanshouxigong, Dongsi, dan Nongzhanguan.
